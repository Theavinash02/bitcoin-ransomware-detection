# Bitcoin Ransomware Detection

Classifying Bitcoin addresses as ransomware-associated or benign using transaction graph
features from the BitcoinHeist dataset (2.9M addresses, 8 features, 29 ransomware families +
a benign `white` class).

Bitcoin has been widely used by ransomware operators to collect payments anonymously. Predicting
which addresses belong to which ransomware family (or to none) from nothing but transaction
graph statistics is a classification problem, but the label distribution is heavily skewed: the
overwhelming majority of addresses are benign, and several ransomware families are represented
by only a handful of known addresses. That imbalance — and being honest about what it does to a
naive accuracy metric — is the throughline of this notebook.

**Contents:**
1. Exploratory data analysis
2. Preprocessing — rare-class handling and a stratified split
3. Baseline model comparison (Logistic Regression, Decision Tree, Random Forest) with
   cross-validation
4. Class-imbalance correction with SMOTE, evaluated with a leakage-safe cross-validation protocol
5. Final held-out test-set evaluation
6. SHAP interpretation of the trained model
7. Discussion and limitations

This is a research/learning project built on a public benchmark dataset — nothing here claims
real-time ransomware detection or production readiness.

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, f1_score, classification_report, confusion_matrix
)

from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline

import shap

import warnings
warnings.filterwarnings('ignore')

RANDOM_STATE = 42
```

## 2. Preprocessing

### Handling the ultra-rare families

SMOTE (used later) needs enough neighbors within a class to interpolate from, and a class with
a single sample can't even be split into train/test, let alone oversampled. Rather than silently
dropping these addresses, every family with fewer than 10 total samples is grouped into a single
`rare_family` bucket before splitting. This is a real loss of resolution — `montrealSam` and
`montrealComradeCircle` become indistinguishable — but it's an honest, documented trade-off
rather than a silent exclusion of data.

```python
rare_labels = label_counts[label_counts < 10].index.tolist()
df['label_grouped'] = df['label'].where(~df['label'].isin(rare_labels), 'rare_family')

print(f"{len(rare_labels)} families grouped into 'rare_family' "
      f"({label_counts[rare_labels].sum()} rows total):")
print(label_counts[rare_labels])
print(f"\nClasses before grouping: {df['label'].nunique()}")
print(f"Classes after grouping:  {df['label_grouped'].nunique()}")
```

### Stratified train/test split

The dataset is 98.6% one class. An unstratified `train_test_split` can, by chance, leave a
already-rare family with proportionally fewer (or zero) samples in one split than the other —
a real methodological weakness on data this skewed. `stratify=y` fixes that by preserving each
class's proportion in both the training and test sets.

```python
X = df[['year', 'day', 'length', 'weight', 'count', 'looped', 'neighbors', 'income']]
y = df['label_grouped']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

print(f"Train: {X_train.shape[0]:,} rows   Test: {X_test.shape[0]:,} rows")
y_train.value_counts()
```

## 3. Baseline Model Comparison

Three classifiers are compared on identical features and identical folds, using 5-fold
`StratifiedKFold` cross-validation on the training set: **Logistic Regression** (with feature
scaling — it's the only one of the three that needs it), **Decision Tree**, and **Random Forest**.
Each fold reports accuracy, weighted F1, and macro F1; the table below shows the mean ± standard
deviation across folds. No SMOTE is applied here — this is a comparison of raw algorithm
performance on the actual, imbalanced training distribution, to establish which model is worth
carrying forward before adding any imbalance correction.

**A note on the Random Forest's tree size:** left fully unconstrained, a Random Forest grown on
2.3M+ rows produces individual trees with over 100,000 nodes each. That's large enough that
`TreeExplainer` (used later, in Section 6) can't build its internal per-tree representation
without running out of memory. `min_samples_leaf=25` and `max_depth=20` bound the trees to a
size that's actually explainable — this is also standard regularization practice on a dataset
this large, and is applied consistently everywhere Random Forest appears below (this comparison,
the SMOTE pipeline, and the final model), so it's not something bolted on only for SHAP to work.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
scoring = ['accuracy', 'f1_weighted', 'f1_macro']

candidate_models = {
    'Logistic Regression': Pipeline([
        ('scaler', StandardScaler()),
        ('clf', LogisticRegression(max_iter=1000, random_state=RANDOM_STATE)),
    ]),
    'Decision Tree': DecisionTreeClassifier(random_state=RANDOM_STATE),
    'Random Forest': RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1, min_samples_leaf=25, max_depth=20),
}

# cross_validate itself runs folds sequentially (n_jobs=1): with 2.9M rows and limited RAM,
# parallelizing across folds would keep several full copies of the data and several full
# Random Forests in memory at once. Each RandomForestClassifier still parallelizes its own
# tree-building internally (n_jobs=-1) via shared-memory threads, which doesn't multiply
# memory the way process-level fold parallelism would.
cv_results = {}
for name, model in candidate_models.items():
    scores = cross_validate(model, X_train, y_train, cv=cv, scoring=scoring, n_jobs=1)
    cv_results[name] = scores
    print(f"{name}: done")
```

```python
summary_rows = []
for name, scores in cv_results.items():
    summary_rows.append({
        'Model': name,
        'Accuracy': f"{scores['test_accuracy'].mean():.4f} \u00b1 {scores['test_accuracy'].std():.4f}",
        'Weighted F1': f"{scores['test_f1_weighted'].mean():.4f} \u00b1 {scores['test_f1_weighted'].std():.4f}",
        'Macro F1': f"{scores['test_f1_macro'].mean():.4f} \u00b1 {scores['test_f1_macro'].std():.4f}",
    })

baseline_comparison = pd.DataFrame(summary_rows).set_index('Model')
baseline_comparison
```

**Why Random Forest:** the comparison doesn't produce the clean "Random Forest wins everything"
result a tutorial would suggest — on point estimates alone, Decision Tree actually edges out
Random Forest on both weighted F1 (0.9827 vs 0.9808) and macro F1 (0.2209 vs 0.1044). That's
worth stating plainly rather than glossing over. But the Decision Tree's macro F1 also carries
roughly 8x the fold-to-fold variance of Random Forest's (±0.0160 vs ±0.0021): a single tree's
apparent edge on the rare classes is unstable, driven by which handful of rare-class rows happen
to land in a given fold, rather than a reliably learned pattern. Random Forest's bagging trades a
bit of raw point-estimate macro F1 for materially more consistent behavior across splits, and its
ensemble structure is what makes `TreeExplainer` straightforward to apply in Section 6. Random
Forest is carried forward on that basis — stability and explainability infrastructure, not
because it swept every metric in this table.

## 4. Class-Imbalance Correction with SMOTE

Random Forest carries forward as the model of choice. This section adds SMOTE to address the
imbalance quantified in Section 1, and re-runs the same 5-fold cross-validation protocol to
check whether the correction actually helps before committing to it for the final model.

**Avoiding leakage:** SMOTE is wrapped together with the classifier inside a single
`imblearn` pipeline, and that pipeline — not a pre-resampled dataset — is what's passed to
`cross_validate`. This matters: if SMOTE were applied once to the whole training set before
cross-validation, synthetic points derived from a validation fold's neighbors could leak into
that same fold's training data. Inside the pipeline, SMOTE only ever sees and resamples the
training portion of each individual fold; the held-out portion of every fold stays untouched and
at its real distribution.

**Oversampling target:** balancing every class up to the size of `white` would inflate the
training set from roughly 2.3M rows to well over 80M across 19 classes — infeasible for a Random
Forest on this project's timeline and not necessary to demonstrate the correction. Instead, SMOTE
oversamples any class *below* a fixed cap up to that cap, and leaves already-larger classes (e.g.
`paduaCryptoWall`, `montrealCryptoLocker`) untouched. Because the pipeline is refit on a
different subset of data each fold, the resampling target is computed as a function of whatever
class counts are actually present in that fold's training data, rather than a static
dictionary computed once — a static target derived from the full training set could ask SMOTE
to *downsample* a class that happens to be smaller within a given fold, which SMOTE cannot do.

```python
CAP = 5000

# k_neighbors must be safe for the smallest class in a single CV fold's training portion,
# not just the full training set - each fold trains on roughly (n_splits-1)/n_splits of it.
min_class_count = y_train.value_counts().min()
worst_case_fold_count = int(min_class_count * (cv.get_n_splits() - 1) / cv.get_n_splits())
k_neighbors = max(1, min(5, worst_case_fold_count - 1))

print(f"Smallest training class: {y_train.value_counts().idxmin()} ({min_class_count} samples)")
print(f"Estimated worst-case count within a single CV fold: {worst_case_fold_count}")
print(f"k_neighbors used for SMOTE: {k_neighbors}")

def smote_sampling_strategy(y_fold):
    """Oversample any class below CAP up to CAP; never touch 'white' or already-large classes."""
    counts = pd.Series(y_fold).value_counts()
    return {cls: max(cnt, CAP) for cls, cnt in counts.items() if cls != 'white'}

smote_rf_pipeline = ImbPipeline([
    ('smote', SMOTE(sampling_strategy=smote_sampling_strategy,
                     k_neighbors=k_neighbors, random_state=RANDOM_STATE)),
    ('rf', RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1, min_samples_leaf=25, max_depth=20)),
])
```

```python
# Sequential folds again, for the same memory reason as Section 3's CV loop.
smote_cv_results = cross_validate(
    smote_rf_pipeline, X_train, y_train, cv=cv, scoring=scoring, n_jobs=1
)

print("SMOTE + Random Forest — 5-fold CV results:")
print(f"Accuracy:    {smote_cv_results['test_accuracy'].mean():.4f} \u00b1 {smote_cv_results['test_accuracy'].std():.4f}")
print(f"Weighted F1: {smote_cv_results['test_f1_weighted'].mean():.4f} \u00b1 {smote_cv_results['test_f1_weighted'].std():.4f}")
print(f"Macro F1:    {smote_cv_results['test_f1_macro'].mean():.4f} \u00b1 {smote_cv_results['test_f1_macro'].std():.4f}")
```

## 5. Final Evaluation

The SMOTE + Random Forest pipeline is refit once on the entire training set and evaluated on the
held-out test set — the real, untouched distribution that has not been resampled, seen during
cross-validation, or used to make any modeling decision. This is the project's single reported
result.

```python
final_pipeline = ImbPipeline([
    ('smote', SMOTE(sampling_strategy=smote_sampling_strategy,
                     k_neighbors=k_neighbors, random_state=RANDOM_STATE)),
    ('rf', RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1, min_samples_leaf=25, max_depth=20)),
])
final_pipeline.fit(X_train, y_train)

y_pred = final_pipeline.predict(X_test)

test_accuracy = accuracy_score(y_test, y_pred)
test_f1_weighted = f1_score(y_test, y_pred, average='weighted')
test_f1_macro = f1_score(y_test, y_pred, average='macro')

print("Test-set results:")
print(f"Accuracy:    {test_accuracy:.4f}")
print(f"Weighted F1: {test_f1_weighted:.4f}")
print(f"Macro F1:    {test_f1_macro:.4f}")
print()
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
class_labels = sorted(y_test.unique())
cm = confusion_matrix(y_test, y_pred, labels=class_labels)

plt.figure(figsize=(12, 10))
sns.heatmap(cm, annot=False, cmap='Blues', xticklabels=class_labels, yticklabels=class_labels)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix — Final Model (held-out test set)')
plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

**Results interpretation:** the final model reaches 98.65% accuracy, 0.9803 weighted F1, and
0.1150 macro F1 on the held-out test set — consistent with the 0.1246 ± 0.0145 macro F1 seen
during cross-validation, so the test-set number isn't a lucky or unlucky single draw.

The classification report shows a specific, honest pattern: several families that have thousands
of test samples and precision at or near 1.00 — `montrealCryptoLocker` (1863 samples, precision
1.00), `paduaCryptoWall` (2478 samples, precision 0.80), `princetonLocky` (1325 samples, precision
1.00) — collapse to near-zero recall (0.00, 0.01, 0.03 respectively) and therefore near-zero F1.
The model rarely calls these families, but when it does, it's usually right. This is a direct,
measurable cost of the `max_depth=20` / `min_samples_leaf=25` constraint from Section 3: those
limits were necessary to keep the Random Forest's trees small enough for `TreeExplainer` to run
in Section 6 without exhausting this machine's memory, but they also stop the trees from carving
out the small, precise decision regions that these minority classes need to be *found*, not just
correctly labeled when stumbled upon. SMOTE's oversampling helps the model learn that these
classes exist (precision is high, not zero), but shallower trees still can't route enough test
cases to them. This is a genuine, non-hidden trade-off between interpretability and raw recall on
rare classes — not a claim that SMOTE "didn't work."

## 6. SHAP Interpretation

`TreeExplainer` computes exact Shapley values for tree ensembles with no special setup, so it's
applied directly to the fitted Random Forest inside `final_pipeline` (SMOTE is a resampling step
that only acts during `.fit` — at prediction/explanation time, data flows through to the forest
unchanged, so the raw test features can be explained directly).

**Sample size:** SHAP values are computed on a random sample of 8,000 test rows rather than the
full 583,340-row test set. `TreeExplainer`'s cost scales with the number of rows explained, and
a 2.9M-address dataset gains nothing from an exhaustive pass — 8,000 rows is enough for stable
global feature-importance rankings and a couple of representative individual explanations.

Because this is a 19-class problem, SHAP produces a separate set of values per class. Rather
than showing 19 summary plots, this section compares the two that matter most for the story:
`white` (benign) and `paduaCryptoWall` (the largest ransomware family in the dataset).

```python
rf_model = final_pipeline.named_steps['rf']

X_shap_sample = X_test.sample(n=8000, random_state=RANDOM_STATE)

explainer = shap.TreeExplainer(rf_model)
shap_values = explainer(X_shap_sample)

print("shap_values.values shape:", shap_values.values.shape)
print("Classes (order matches last axis):", list(rf_model.classes_))
```

```python
white_idx = list(rf_model.classes_).index('white')
shap.summary_plot(shap_values.values[:, :, white_idx], X_shap_sample, show=False)
plt.title("SHAP Summary \u2014 'white' (benign) class")
plt.tight_layout()
plt.show()
```

```python
ransom_class = 'paduaCryptoWall'
ransom_idx = list(rf_model.classes_).index(ransom_class)
shap.summary_plot(shap_values.values[:, :, ransom_idx], X_shap_sample, show=False)
plt.title(f"SHAP Summary \u2014 '{ransom_class}' class")
plt.tight_layout()
plt.show()
```

```python
sample_true = y_test.loc[X_shap_sample.index]
sample_pred = rf_model.predict(X_shap_sample)

correct_white_mask = (sample_true.values == 'white') & (sample_pred == 'white')
correct_ransom_mask = (sample_true.values == ransom_class) & (sample_pred == ransom_class)

white_pos = np.where(correct_white_mask)[0][0]
ransom_candidates = np.where(correct_ransom_mask)[0]
ransom_pos = ransom_candidates[0] if len(ransom_candidates) > 0 else np.where(sample_true.values == ransom_class)[0][0]

print(f"'white' example -> row position {white_pos}")
print(f"'{ransom_class}' example -> row position {ransom_pos}")
```

```python
shap.plots.force(shap_values[white_pos, :, white_idx], matplotlib=True, show=False)
plt.title("Force plot \u2014 correctly predicted 'white' address")
plt.tight_layout()
plt.show()
```

```python
shap.plots.force(shap_values[ransom_pos, :, ransom_idx], matplotlib=True, show=False)
plt.title(f"Force plot \u2014 '{ransom_class}' address")
plt.tight_layout()
plt.show()
```

**SHAP interpretation:** for both classes shown, `year` is the single dominant feature by a wide
margin — its SHAP values span roughly ±0.3 for `white` and up to +0.11 for `paduaCryptoWall`,
noticeably wider than any other feature's spread. For `white`, `length` and `count` follow, with
high transaction-chain length pushing predictions toward benign. For `paduaCryptoWall`, `day` and
`neighbors` are the next-most-influential features after `year`.

That ranking is a genuinely useful, slightly uncomfortable finding rather than a confirmation of
the domain story this project expected going in: `count` and `neighbors` — the features that
should, per the payment-consolidation intuition, be the clearest ransomware signal — rank *behind*
the purely temporal features `year` and `day` for `paduaCryptoWall`. The likely explanation is
that `paduaCryptoWall` was a real ransomware campaign active within a specific, narrow window of
the 2011–2018 dataset, so the model may be substantially keying on *when* an address was active
rather than learning a transaction-shape fingerprint that would generalize to a ransomware family
outside this dataset. That's the honest limitation of this interpretation: SHAP shows what the
model relies on, and here that includes a feature (`year`) that plausibly encodes "which known
campaign was running," which is a much narrower and more dataset-specific signal than genuine
behavioral detection.

## 7. Discussion and Limitations

**What this shows:** accuracy alone was a misleading metric on a 98.6%-imbalanced dataset — the
baseline comparison and macro F1 numbers above make that explicit rather than hiding behind a
single headline accuracy figure. SMOTE, applied without leaking information across folds or
into the test set, addresses part of that imbalance; SHAP then explains *why* the resulting model
makes the decisions it does, rather than only reporting that it does.

**Limitations, stated plainly:**
- This is a public benchmark dataset covering 2011–2018 Bitcoin activity, not a live monitoring
  feed. Nothing here claims real-time ransomware detection or production readiness.
- Grouping 11 ultra-rare families into a single `rare_family` bucket is a real loss of
  resolution — those families remain indistinguishable from each other in this analysis.
- SMOTE interpolates *within* existing minority-class structure; it cannot manufacture genuinely
  new information about families that had only a handful of real addresses to begin with, and the
  most extreme rare classes are likely to remain difficult even after correction.
- The `max_depth`/`min_samples_leaf` constraints needed to make the Random Forest small enough
  for `TreeExplainer` to run measurably hurt recall on several well-represented ransomware
  families (Section 5) — precision on those families stays high, but the model finds far fewer
  of them than an unconstrained forest would. Interpretability and raw recall were in direct
  tension here, and interpretability was chosen deliberately, but it has a real cost.
- SHAP explains association and feature attribution learned by this specific model — it is not a
  causal account of *why* ransomware addresses behave the way they do. Section 6 also surfaces a
  concrete instance of this: the model's reliance on `year` for `paduaCryptoWall` may reflect a
  narrow, campaign-specific temporal fingerprint rather than a generalizable behavioral signal.

**A natural extension** would be a complementary probabilistic model — for example, a Bayesian
Network over the same 8 features — that offers an explicit, human-auditable conditional
independence structure and a full joint distribution, as a counterpart to the Random Forest's
post-hoc, per-prediction SHAP explanations. That comparison is left for a future update to this
project.
